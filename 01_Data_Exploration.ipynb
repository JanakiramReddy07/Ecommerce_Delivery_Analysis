````python
import pandas as pd

df = pd.read_csv("DATA/Raw Data/DataCoSupplyChainDataset.csv", encoding="latin1")

df.head()
````

````python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
````

````python
df.columns.tolist()
````

````python
df.info()
````

````python
missing_values = df.isnull().sum()

missing_values = missing_values[missing_values > 0]

print(missing_values)
````

````python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_percentage = missing_percentage[missing_percentage > 0]

print(missing_percentage.round(2))
````

````python
df = df.drop(columns=["Product Description", "Order Zipcode"])

df.shape
````

````python
df.isnull().sum()[df.isnull().sum() > 0]
````

````python
df["Customer Lname"] = df["Customer Lname"].fillna("Unknown")
````

````python
df.isnull().sum()[df.isnull().sum() > 0]
````

````python
df.duplicated().sum()
````

````python
df.dtypes
````

````python
df[["order date (DateOrders)", "shipping date (DateOrders)"]].head()
````

````python
df["order date (DateOrders)"] = pd.to_datetime(
    df["order date (DateOrders)"],
    format="%m/%d/%Y %H:%M"
)
````

````python
df["shipping date (DateOrders)"] = pd.to_datetime(
    df["shipping date (DateOrders)"],
    format="%m/%d/%Y %H:%M"
)
````

````python
df[["order date (DateOrders)", "shipping date (DateOrders)"]].dtypes
````

````python
df["Delivery_Delay"] = (
    df["Days for shipping (real)"] -
    df["Days for shipment (scheduled)"]
)
````

````python
df[[
    "Days for shipping (real)",
    "Days for shipment (scheduled)",
    "Delivery_Delay"
]].head(10)
````

````python
df["Delivery_Performance"] = df["Delivery_Delay"].apply(
    lambda x: "Early" if x < 0 else ("Late" if x > 0 else "On Time")
)

df["Delivery_Performance"].value_counts()
````

````python
delivery_percentage = (
    df["Delivery_Performance"].value_counts(normalize=True) * 100
).round(2)

print(delivery_percentage)
````

````python
import matplotlib.pyplot as plt

delivery_counts = df["Delivery_Performance"].value_counts()

delivery_counts.plot(kind="bar")

plt.title("E-commerce Delivery Performance")
plt.xlabel("Delivery Performance")
plt.ylabel("Number of Orders")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
````

````python
delivery_counts = df["Delivery_Performance"].value_counts()

delivery_counts.plot(kind="bar")

plt.title("E-commerce Delivery Performance")
plt.xlabel("Delivery Performance")
plt.ylabel("Number of Orders")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
````

````python
df.groupby("Shipping Mode")["Delivery_Performance"].value_counts()
````

````python
late_delivery_percentage = (
    pd.crosstab(
        df["Shipping Mode"],
        df["Delivery_Performance"],
        normalize="index"
    ) * 100
).round(2)

late_delivery_percentage
````

````python
late_delivery_percentage.plot(kind="bar", stacked=True)

plt.title("Delivery Performance by Shipping Mode")
plt.xlabel("Shipping Mode")
plt.ylabel("Percentage of Orders")
plt.xticks(rotation=0)
plt.legend(title="Delivery Performance")
plt.tight_layout()
plt.show()
````

````python
category_delay = df.groupby("Category Name")["Delivery_Delay"].mean().sort_values(ascending=False)

category_delay
````

````python
top_10_delay = category_delay.head(10)

top_10_delay.plot(kind="bar")

plt.title("Top 10 Product Categories by Average Delivery Delay")
plt.xlabel("Product Category")
plt.ylabel("Average Delivery Delay (Days)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
````

````python
region_delay = (
    df.groupby("Order Region")["Delivery_Delay"]
    .mean()
    .sort_values(ascending=False)
)

region_delay
````

````python
top_10_regions = region_delay.head(10)

top_10_regions.plot(kind="bar")

plt.title("Top 10 Regions by Average Delivery Delay")
plt.xlabel("Order Region")
plt.ylabel("Average Delivery Delay (Days)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
````

````python
market_delay = (
    df.groupby("Market")["Delivery_Delay"]
    .mean()
    .sort_values(ascending=False)
)

market_delay
````

````python
market_delay.plot(kind="bar")

plt.title("Average Delivery Delay by Market")
plt.xlabel("Market")
plt.ylabel("Average Delivery Delay (Days)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
````

````python
top_customers = (
    df.groupby("Customer Id")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_customers
````

````python
top_customers.plot(kind="bar")

plt.title("Top 10 Customers by Total Sales")
plt.xlabel("Customer ID")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
````

````python
category_sales = (
    df.groupby("Category Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

category_sales
````

````python
top_10_category_sales = category_sales.head(10)

top_10_category_sales.plot(kind="bar")

plt.title("Top 10 Product Categories by Total Sales")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
````

````python
category_profit = (
    df.groupby("Category Name")["Order Profit Per Order"]
    .sum()
    .sort_values(ascending=False)
)

category_profit
````

````python
top_10_category_profit = category_profit.head(10)

top_10_category_profit.plot(kind="bar")

plt.title("Top 10 Product Categories by Total Profit")
plt.xlabel("Product Category")
plt.ylabel("Total Profit")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
````

````python
monthly_sales = (
    df.groupby(df["order date (DateOrders)"].dt.to_period("M"))["Sales"]
    .sum()
)

monthly_sales
````

````python
total_orders = len(df)

total_sales = df["Sales"].sum()

total_profit = df["Order Profit Per Order"].sum()

average_delivery_delay = df["Delivery_Delay"].mean()

late_delivery_percentage = (
    df["Delivery_Performance"].eq("Late").mean() * 100
)

print("Total Records:", total_orders)
print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Average Delivery Delay:", round(average_delivery_delay, 2), "days")
print("Late Delivery Percentage:", round(late_delivery_percentage, 2), "%")
````
