```python
import pandas as pd

df = pd.read_csv(
    "../DATA/Raw Data/DataCoSupplyChainDataset.csv",
    encoding="latin1"
)

print("Dataset loaded successfully!")

df.head()
```

```python
print("Dataset Shape:", df.shape)

df.info()
```

```python
missing_values = df.isnull().sum()

print(missing_values[missing_values >0])
```

```python
missing_summary = pd.DataFrame({
    'Missing Values': df.isnull().sum(),
    'Missing Percentage': (df.isnull().sum() / len(df)) * 100
})

missing_summary = missing_summary[missing_summary['Missing Values'] > 0]

print(missing_summary)
```

```python
df.drop(columns=['Product Description'], inplace=True)

print("Product Description column removed successfully!")

df.shape
```

```python
df['Customer Lname'] = df['Customer Lname'].fillna('Unknown')

print("Missing values:", df['Customer Lname'].isnull().sum())
```

```python
print("Order Zipcode data type:")
print(df['Order Zipcode'].dtype)

print("\nSample Order Zipcode values:")
print(df['Order Zipcode'].head(10))

print("\nTotal missing values:")
print(df['Order Zipcode'].isnull().sum())
```

```python
print(df['Order Zipcode'].dropna().head(10))

print("\nAvailable values:",df['Order Zipcode'].count())

print("Unique zipcodes:",df['Order Zipcode'].nunique())
```

```python
df.drop(columns=['Order Zipcode'], inplace=True)

print("Order Zipcode column removed successfully!")

print("Updated dataset shape:", df.shape)
```

```python
print("Data type:", df['Customer Zipcode'].dtype)

print("\nMissing values:",
      df['Customer Zipcode'].isnull().sum())

print("\nSample values:")
print(df['Customer Zipcode'].head(10))
```

```python
duplicates = df.duplicated().sum()

print("Total duplicate records:", duplicates)
```

```python
print ( df.dtypes)
```

```python
print("Order Date:")
print(df['order date (DateOrders)'].head())

print("\nShipping Date:")
print(df['shipping date (DateOrders)'].head())
```

```python
df['order date (DateOrders)'] = pd.to_datetime(
    df['order date (DateOrders)'],
    format='%m/%d/%Y %H:%M'
)

df['shipping date (DateOrders)'] = pd.to_datetime(
    df['shipping date (DateOrders)'],
    format='%m/%d/%Y %H:%M'
)

print("Date columns converted successfully!")

print(df[['order date (DateOrders)',
          'shipping date (DateOrders)']].dtypes)
```

```python
print(df[['order date (DateOrders)',
          'shipping date (DateOrders)']].dtypes)
```

```python
print("Negative Sales:", (df['Sales'] < 0).sum())

print("Negative Quantity:",
      (df['Order Item Quantity'] < 0).sum())

print("Negative Product Price:",
      (df['Product Price'] < 0).sum())

print("Negative Order Profit:",
      (df['Order Profit Per Order'] < 0).sum())
```

```python
invalid_dates = (
    df['shipping date (DateOrders)'] <
    df['order date (DateOrders)']
)

print("Invalid delivery date records:", invalid_dates.sum())
```

```python
print("Remaining missing values:")
print(df.isnull().sum()[df.isnull().sum() > 0])

print("\nFinal dataset shape:", df.shape)
```

```python
df.to_csv(
    "../DATA/Processed/Cleaned_Ecommerce_Dataset.csv",
    index=False)
print("Cleaned dataset saved successfully!")
```
